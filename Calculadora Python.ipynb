```python

print("Olá e bem-vindo à calculadora em Python!")


# 1. Funções matemáticas

def soma(a, b):
    return a + b


def subtracao(a, b):
    return a - b


def multiplicacao(a, b):
    return a * b


def divisao(a, b):
    if b == 0:
        return None
    return a / b


# 2. Laço para realizar os cálculos

while True:

    # 3. Menu e opção escolhida

    print("\nEscolha a operação desejada:")
    print("1 - Soma")
    print("2 - Subtração")
    print("3 - Multiplicação")
    print("4 - Divisão")
    print("0 - Sair")

    try:
        operacao = int(input("Digite a operação desejada: "))

    except ValueError:
        print("Opção inválida. Digite um número inteiro.")
        continue

    if operacao == 0:
        print("Obrigado por utilizar a calculadora!")
        break

    elif operacao not in (1, 2, 3, 4):
        print("Opção inválida. Escolha uma opção de 1 a 4.")
        continue

    # 4. Recebe os números

    try:
        numero1 = float(input("Digite o primeiro número: "))
        numero2 = float(input("Digite o segundo número: "))

    except ValueError:
        print("Por favor, digite números válidos.")
        continue

    # 5. Executa a operação escolhida

    if operacao == 1:
        resultado = soma(numero1, numero2)
        nome_operacao = "soma"

    elif operacao == 2:
        resultado = subtracao(numero1, numero2)
        nome_operacao = "subtração"

    elif operacao == 3:
        resultado = multiplicacao(numero1, numero2)
        nome_operacao = "multiplicação"

    elif operacao == 4:
        resultado = divisao(numero1, numero2)

        if resultado is None:
            print("Não é possível dividir por zero.")
            continue

        nome_operacao = "divisão"

    # 6. Exibe o resultado utilizando f-string

    print(f"O resultado da {nome_operacao} é: {round(resultado, 2)}")

    # 7. Pergunta se o usuário deseja realizar outro cálculo

    continuar = input("Deseja realizar outro cálculo? (s/n): ").strip().lower()
     # 8.Agradecimento por utilizar a calculadora
    if continuar != "s":
        print("Obrigado por utilizar a calculadora em Python!")
        break


```
